```python
import pyvisa
import numpy as np
from pyKeithley import myKeithley

#import sys
#sys.append.path()
```

```python
rm = pyvisa.ResourceManager()
addr = rm.list_resources() # USB0 is the top USB port in RPy, USB1 is the bottom one
addr
```

```python
ps = myKeithley()
ps.connect(addr[0])
```

```python
ps.switch_on_output()
```

```python
# initialize the Keithley settings:
ps.set_tensions(tension_CH1=7,tension_CH2=4,tension_CH3=4)
ps.set_current_limitations(current_CH1=0.8,current_CH2=1.6,current_CH3=0.15)
```

```python
ps.switch_on_output()
```

```python
ps.set_tensions(tension_CH1=7,tension_CH2=4,tension_CH3=4)
ps.set_current_limitations(current_CH1=0.8,current_CH2=1.6,current_CH3=0.15)
```

```python
ps.set_single_tension(Ch_idx=3,tension_value=3)
```

```python
ps.print_settings()
```

```python
ps.switch_off_output()
```

```python
ps.close()
```
